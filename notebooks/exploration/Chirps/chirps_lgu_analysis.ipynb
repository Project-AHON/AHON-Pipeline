```python
def prepare_city_dataset(gdf, lgu_type):
    df = gdf.copy()

    df["source_psgc_code"] = df["psgc_id"]

    df["canonical_psgc_code"] = df["psgc_id"]

    df["canonical_lgu_name"] = df["ADM3_EN"]

    df["reconciliation_status"] = "source_verified"

    df["lgu_type"] = lgu_type

    return df[
        [
            "ADM3_EN",
            "ADM3_PCODE",
            "source_psgc_code",
            "canonical_psgc_code",
            "canonical_lgu_name",
            "reconciliation_status",
            "lgu_type",
            "geometry",
        ]
    ]
```

```python
component_cities_master = prepare_city_dataset(
    component_cities,
    "Component City"
)

huc_master = prepare_city_dataset(
    huc,
    "Highly Urbanized City"
)

independent_cities_master = prepare_city_dataset(
    independent_component_cities,
    "Independent Component City"
)
```

```python
def prepare_city_dataset(gdf, lgu_type):
    df = gdf.copy()

    df["source_psgc_code"] = df["psgc_id"]
    df["canonical_psgc_code"] = df["psgc_id"]
    df["canonical_lgu_name"] = df["ADM3_EN"]
    df["reconciliation_status"] = "source_verified"
    df["lgu_type"] = lgu_type

    return df[
        [
            "ADM3_EN",
            "ADM3_PCODE",
            "source_psgc_code",
            "canonical_psgc_code",
            "canonical_lgu_name",
            "reconciliation_status",
            "lgu_type",
            "geometry",
        ]
    ]
```

```python
component_cities_master = prepare_city_dataset(
    component_cities,
    "Component City"
)

huc_master = prepare_city_dataset(
    huc,
    "Highly Urbanized City"
)

independent_cities_master = prepare_city_dataset(
    independent_component_cities[
        independent_component_cities["ADM3_EN"].notna()
    ],
    "Independent Component City"
)
```
